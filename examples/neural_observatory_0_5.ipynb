## 1. Neural Execution Tracer & Tensor Flow Explorer
Inspect how tensors travel through layers during a real forward pass without fabricating execution paths or modifying model state.

```python
class ResidualBlock(nn.Module):
    def __init__(self, channels: int):
        super().__init__()
        self.conv1 = nn.Conv2d(channels, channels, 3, padding=1)
        self.bn1 = nn.BatchNorm2d(channels)
        self.relu = nn.ReLU()
        self.conv2 = nn.Conv2d(channels, channels, 3, padding=1)
        self.bn2 = nn.BatchNorm2d(channels)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        res = x
        out = self.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        return self.relu(out + res)

model = nn.Sequential(
    nn.Conv2d(3, 16, 3, padding=1),
    ResidualBlock(16),
    nn.AdaptiveAvgPool2d((1, 1)),
    nn.Flatten(),
    nn.Linear(16, 10)
)

x = torch.randn(1, 3, 32, 32)
trace = sd.trace(model, input=x)
flow_viz = sd.flow(trace)
flow_viz
```

## 2. Transformer Observatory & Attention Rollout
Mathematically sound cumulative rollout across transformer layers ($R = A'_L @ \dots @ A'_1$) with row stochasticity and residual blending.

```python
tokens = ["[CLS]", "The", "neural", "observatory", "reveals", "information", "flow", "[SEP]"]
layers_count = 6
heads_count = 4
seq_len = len(tokens)

# Synthesize attention weights for 6 layers
raw_attn = torch.softmax(torch.randn(layers_count, heads_count, seq_len, seq_len), dim=-1)

rollout_viz = sd.attention_rollout(raw_attn, tokens=tokens, include_residual=True)
rollout_viz
```

## 3. Advanced Explainability & Completeness Diagnostics
Compare SmoothGrad noise reduction against Integrated Gradients with exact path completeness error verification.

```python
classifier = nn.Sequential(nn.Linear(10, 20), nn.ReLU(), nn.Linear(20, 2))
sample_x = torch.randn(1, 10)

# Integrated Gradients
ig_viz = sd.explain(classifier, input=sample_x, target=0, method="integrated_gradients", steps=30)
print("IG Completeness Error:", ig_viz.data["diagnostics"]["completeness_error"])

# SmoothGrad
sg_viz = sd.explain(classifier, input=sample_x, target=0, method="smoothgrad", samples=20, seed=42)
sg_viz
```

## 4. Gradient Health & Numerical Stability Intelligence
Inspect exact parameter gradient norms without mutation, detecting vanishing or exploding gradients and tied weights.

```python
loss = classifier(sample_x).sum()
loss.backward()

health_viz = sd.gradient_health(classifier)
health_viz
```
